# wap to find the temperture in celcius from the farenheit input

$C = (F-32)5/9$

```python
def celcius(farenheit):
    c = (farenheit-32) *(5/9)
    print(f"for farenheit{farenheit}, Temperature in celcius is {c}")

celcius(98) # calling of the function
celcius(102) # calling of the function
```

```python
def celcius(farenheit):
    c = (farenheit-32) *(5/9)
    print(f"for farenheit{farenheit}, Temperature in celcius is {round(c,2)}")

celcius(98) # calling of the function
celcius(102) # calling of the function
```

# wap to print whether a 3 digit number is an armstrong number

```python
def armstrong():
    num = int(input("enter any number to check whether armstorng or not: "))

    length = len(str(num))
    result =0
    for i in str(num):
        result= result + (int(i)**length)

    if result == num:
        print("this number",num,"is an armstrong number !")
    else:
        print ("oh!, this number is not an armstrong number.")

armstrong()
    
```

```python
num = int(input("enter any number to check whether armstorng or not: "))

length = len(str(num))
result =0
for i in str(num):
    result= result + (int(i)**length)

if result == num:
    print("this number",num,"is an armstrong number !")
else:
    print ("oh!, this number is not an armstrong number.")
```

# factorial

```python
# 5! = 5x4x3x2x1 =120
# 3! = 3x2x1 =6 
def fact (n):
    if n==1:
        return 1
    else:
        product = n * fact(n-1)
        return(product)
fact(3)
```

```python
maximum=0
list1 =[50,40,60,20,5,7,80,6,12,8,6,80.5,65]
counter =1
for i in list1:
    if i > maximum:
        maximum =i
        print(f"maximum in iteration{counter} is {maximum}")
        counter+=i
print("\n---------------------\nFinal Result is: ",maximum)
```
